```python
Q1 = df['actual_price'].quantile(0.25)
Q3 = df['actual_price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

actual_price_outliers = df[
    (df['actual_price'] < lower) |
    (df['actual_price'] > upper)
]

print("Number of Outliers:", len(actual_price_outliers))
actual_price_outliers[['product_name','actual_price']]
```

```python
corr_matrix = df[numeric_cols].corr()

corr_pairs = corr_matrix.unstack()

corr_pairs = corr_pairs[corr_pairs < 1]

strongest_positive = corr_pairs.sort_values(
    ascending=False
).head(1)

print(strongest_positive)
```
